## Import Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

print("Libraries loaded")
```

## Step 2 — Load the CSV



```python
df = pd.read_csv("loan_applications.csv")

print("Dataset loaded")
print("Rows and columns:", df.shape)
df.head()
```

# Step 1 — Central Tendency & Dispersion

### 1A. Mean, Median and Mode of Income


```python
mean_income = df["Income"].mean()
median_income = df["Income"].median()
mode_income = df["Income"].mode()[0]

print(f"Mean Income   : {mean_income:,.2f}")
print(f"Median Income : {median_income:,.2f}")
print(f"Mode Income   : {mode_income:,.2f}")
```

### 1B. Range, Variance and Standard Deviation of Loan Amount



```python
loan_range = df["Loan_Amount"].max() - df["Loan_Amount"].min()
loan_variance = df["Loan_Amount"].var()
loan_std = df["Loan_Amount"].std()

print(f"Minimum Loan Amount : {df['Loan_Amount'].min():,.2f}")
print(f"Maximum Loan Amount : {df['Loan_Amount'].max():,.2f}")
print(f"Range               : {loan_range:,.2f}")
print(f"Variance            : {loan_variance:,.2f}")
print(f"Standard Deviation  : {loan_std:,.2f}")
```

# Step 2 — Probability & Events

### 2A. Probability of loan default



```python
p_default = (df["Default_Status"] == "Yes").mean()
print(f"Probability of Default = {p_default:.4f}")
print(f"Percentage of Default = {p_default*100:.2f}%")
```

### 2B. Contingency table: Default Status vs Credit Score range


```python
bins = [0, 599, 699, 799, 850]
labels = ["<600", "600-699", "700-799", "800+"]

df["Credit_Range"] = pd.cut(
    df["Credit_Score"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

contingency = pd.crosstab(df["Credit_Range"], df["Default_Status"])
print("Contingency Table:")
display(contingency)
```

### 2C. Conditional probability P(Default | Credit Score < 600)



```python
low_credit = df[df["Credit_Score"] < 600]

p_default_given_low = (
    (low_credit["Default_Status"] == "Yes").sum() / len(low_credit)
)

print("Customers with Credit Score < 600:", len(low_credit))
print("Defaults among them:", (low_credit["Default_Status"] == "Yes").sum())
print(f"P(Default | Credit Score < 600) = {p_default_given_low:.4f}")
print(f"Percentage = {p_default_given_low*100:.2f}%")
```

# Step 3 — Distributions & Visualization

### 3A. Histogram of Credit Score with Gaussian curve


```python
x = df["Credit_Score"]

mu = x.mean()
sigma = x.std()

plt.figure(figsize=(8,5))
plt.hist(x, bins=30, density=True, alpha=0.65, edgecolor="black")

xx = np.linspace(x.min(), x.max(), 400)
gaussian = (
    1 / (sigma * np.sqrt(2*np.pi))
    * np.exp(-0.5 * ((xx - mu) / sigma)**2)
)

plt.plot(xx, gaussian, linewidth=2)
plt.title("Credit Score Histogram with Gaussian Curve")
plt.xlabel("Credit Score")
plt.ylabel("Density")
plt.grid(alpha=0.2)
plt.show()

print(f"Mean = {mu:.2f}")
print(f"Standard Deviation = {sigma:.2f}")
```

### 3B. Skewness and Kurtosis of Loan Amount



```python
loan_skew = df["Loan_Amount"].skew()
loan_kurtosis = df["Loan_Amount"].kurt()

print(f"Loan Amount Skewness = {loan_skew:.4f}")
print(f"Loan Amount Kurtosis = {loan_kurtosis:.4f}")

if loan_skew > 0:
    print("Interpretation: positively/right skewed.")
elif loan_skew < 0:
    print("Interpretation: negatively/left skewed.")
else:
    print("Interpretation: approximately symmetric.")
```

### 3C. Q-Q Plot for Income



```python
plt.figure(figsize=(8,5))
stats.probplot(df["Income"], dist="norm", plot=plt)
plt.title("Q-Q Plot of Income")
plt.grid(alpha=0.2)
plt.show()
```

# Step 4 — Linear Algebra Application



```python
vectors = df.loc[:4, ["Income", "Loan_Amount"]].to_numpy(dtype=float)

for i, v in enumerate(vectors, start=1):
    print(f"Customer {i} vector: {v}")
```

### 4A. Dot product between two customer vectors



```python
A = vectors[0]
B = vectors[1]

dot_product = np.dot(A, B)

print("Customer 1 vector:", A)
print("Customer 2 vector:", B)
print(f"Dot product = {dot_product:,.2f}")
```

### 4B. Norm 2 of a customer's financial vector



```python
norm_customer_1 = np.linalg.norm(A)
print(f"L2 norm of Customer 1 = {norm_customer_1:,.2f}")
```

### 4C. Angle between two customer vectors


```python
cos_theta = dot_product / (
    np.linalg.norm(A) * np.linalg.norm(B)
)

cos_theta = np.clip(cos_theta, -1, 1)
angle_degrees = np.degrees(np.arccos(cos_theta))

print(f"cos(theta) = {cos_theta:.6f}")
print(f"Angle between Customer 1 and Customer 2 = {angle_degrees:.2f} degrees")
```
