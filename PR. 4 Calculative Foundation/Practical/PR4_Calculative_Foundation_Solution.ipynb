## 1. Dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("student_performance.csv")
df
```

## Part A – Vector & Matrix Fundamentals

```python
X = df[["Mathematics","Physics","Computer Science","Statistics"]].to_numpy(dtype=float)
students = df["Student"].tolist()

for name, v in zip(students, X):
    print(name, "=", v)
```

### A1. Norm-1 and Norm-2

```python
norm1 = np.sum(np.abs(X), axis=1)
norm2 = np.linalg.norm(X, axis=1)

result = pd.DataFrame({"Student": students, "Norm-1": norm1, "Norm-2": norm2})
result
```

**Interpretation:** Norm-1 is the sum of absolute scores, while Norm-2 is the Euclidean length of the score vector. Larger values indicate a larger overall score-vector magnitude.

### A2. Dot Product and Angle

```python
v1, v2 = X[0], X[1]
dot = np.dot(v1, v2)
angle = np.degrees(np.arccos(dot/(np.linalg.norm(v1)*np.linalg.norm(v2))))
print("Vector 1:", v1)
print("Vector 2:", v2)
print("Dot product:", dot)
print("Angle (degrees):", angle)
```

**Interpretation:** A small angle means the two performance vectors have similar direction across subjects. The positive dot product also indicates strong alignment.

### A3. Cross Product for 3D Selected Subjects

```python
a3 = v1[:3]  # Mathematics, Physics, Computer Science
b3 = v2[:3]
cross = np.cross(a3, b3)
print("a =", a3)
print("b =", b3)
print("a × b =", cross)
print("a · (a × b) =", np.dot(a3, cross))
print("b · (a × b) =", np.dot(b3, cross))
```

**Interpretation:** The cross product creates a vector perpendicular to both selected 3D score vectors. The two near-zero dot products verify perpendicularity (up to floating-point precision).

### A4. Projection

```python
projection = (np.dot(v1,v2)/np.dot(v2,v2))*v2
projection
```

**Interpretation:** The projection is the component of Aarav's score vector that lies in the direction of Bhavya's score vector.

## Part B – Matrix Operations

```python
M = X[:4]  # 4 students × 4 subjects, used for square-matrix operations
print("M =")
print(M)
print("\nM + M =")
print(M + M)
print("\nTranspose M^T =")
print(M.T)
```

**Interpretation:** Matrix addition is element-wise. Transpose swaps rows and columns, changing a 4×4 matrix into another 4×4 matrix.

### B1. Matrix Multiplication

```python
ATA = X.T @ X
print("X^T X =")
print(ATA)
```

**Interpretation:** XᵀX is a 4×4 matrix that captures pairwise products among subject-score columns and is useful in covariance/PCA and linear-algebra calculations.

### B2. Determinant and Inverse

```python
det_M = np.linalg.det(M)
inv_M = np.linalg.inv(M)

print("det(M) =", det_M)
print("\nM^-1 =")
print(inv_M)
print("\nM @ M^-1 ≈")
print(M @ inv_M)
```

**Interpretation:** Because det(M) is non-zero, M is invertible. The product M·M⁻¹ is approximately the identity matrix, allowing for floating-point rounding.

## Part C – Linear Transformations & Geometry

```python
# Simple 2D -> 3D -> 4D dimensionality illustration
print("2D vector:", X[0,:2])
print("3D vector:", X[0,:3])
print("4D vector:", X[0,:4])
```

## Part D – Eigenvalues & Decomposition

```python
cov = np.cov(X, rowvar=False)
eigvals, eigvecs = np.linalg.eigh(cov)
order = np.argsort(eigvals)[::-1]
eigvals = eigvals[order]
eigvecs = eigvecs[:, order]
explained = eigvals/eigvals.sum()*100

print("Covariance matrix:")
print(cov)
print("\nEigenvalues:")
print(eigvals)
print("\nEigenvectors (columns):")
print(eigvecs)
print("\nExplained variance (%):")
print(explained)
```

**Interpretation:** Eigenvalues measure the variance captured by each principal direction. The eigenvector associated with the largest eigenvalue is the first principal direction. Ordering eigenvalues from largest to smallest ranks the directions by explained variance.

### PCA-style dimensionality reduction to 2D

```python
X_centered = X - X.mean(axis=0)
scores2 = X_centered @ eigvecs[:, :2]

plt.figure(figsize=(7,5))
plt.scatter(scores2[:,0], scores2[:,1], s=70)
for i, name in enumerate(students):
    plt.annotate(name, (scores2[i,0], scores2[i,1]), xytext=(5,5), textcoords="offset points")
plt.axhline(0, linewidth=0.7)
plt.axvline(0, linewidth=0.7)
plt.title("Student Performance – First Two Principal Components")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()
```

**Interpretation:** PCA projects the original four-dimensional subject data onto a smaller number of directions while retaining as much variance as possible. PC1 and PC2 provide a compact 2D representation of the students.
